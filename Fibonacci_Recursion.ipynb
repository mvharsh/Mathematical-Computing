# **IMPLEMENTATION OF RECURSION AND RECURRENCES WITH VISUALIZATION**

**AIM:**

To demonstrate recursion and recurrence relations using the Fibonacci sequence and visualize the results using Python.


**ALGORITHM:**

1.	Define a recursive function to compute Fibonacci numbers using the recurrence relation: ( F(n) = F(n-1) + F(n-2), F(0) = 0, F(1) = 1
2.	Write a function to generate a list of Fibonacci numbers for a given number of terms.
3.	Use the `matplotlib` library to plot the Fibonacci sequence.
4.	Run the program to compute and visualize the Fibonacci sequence.


```python
import matplotlib.pyplot as plt

# Recursive function to compute Fibonacci numbers
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# Generate Fibonacci sequence for visualization
def generate_fibonacci_sequence(n):
    sequence = []
    for i in range(n):
        sequence.append(fibonacci(i))
    return sequence

# Plotting the Fibonacci sequence
def plot_fibonacci(sequence):
    plt.figure(figsize=(10, 6))
    plt.plot(sequence, marker='o', linestyle='-', color='b', label='Fibonacci Sequence')
    plt.title("Fibonacci Sequence Visualization", fontsize=16)
    plt.xlabel("Index (n)", fontsize=14)
    plt.ylabel("Fibonacci Number", fontsize=14)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend()
    plt.show()
# Main Function
if __name__ == "__main__":
    n = int(input("Enter n: "))
    print(f"Generating the first {n} terms of the Fibonacci sequence...")
    fibonacci_sequence = generate_fibonacci_sequence(n)
    print(f"Fibonacci Sequence: {fibonacci_sequence}")
    print("Visualizing the Fibonacci sequence...")
    plot_fibonacci(fibonacci_sequence)
```
