# **SOLVING ASSIGNMENT PROBLEMS USING THE HUNGARIAN ALGORITHM**

**AIM:**

To implement the Hungarian Algorithm in Python to solve the Assignment Problem by minimizing the total cost.


**ALGORITHM:**

- **Formulate the Cost Matrix:** Create a matrix representing the cost of assigning tasks to agents.

- **Row Reduction:** Subtract the smallest value in each row from all elements in that row.

- **Column Reduction:** Subtract the smallest value in each column from all elements in that column.

- **Cover Zeros:** Use the minimum number of lines to cover all zeros.

- **Adjust Matrix (if necessary):** If the number of lines < matrix size, adjust by subtracting the smallest uncovered value and adding it to intersections.

- **Repeat Until Optimal Assignment is Found:** Continue until an assignment is found using the zeros.

```python
import numpy as np
from scipy.optimize import linear_sum_assignment

def hungarian_algorithm(cost_matrix):
    # Apply the Hungarian Algorithm
    row_ind, col_ind = linear_sum_assignment(cost_matrix)

    # Calculate the total cost
    total_cost = cost_matrix[row_ind, col_ind].sum()

    # Format assignment output as a list of tuples with standard Python integers
    assignment = [(int(r), int(c)) for r, c in zip(row_ind, col_ind)]

    # Return assignment and total cost
    return assignment, total_cost

# Get user input for cost matrix
def get_user_input():
    n = int(input("Enter the number of agents/tasks: "))
    print("Enter the cost matrix row by row, separating values by spaces:")
    cost_matrix = []
    for i in range(n):
        row = list(map(int, input().split()))
        cost_matrix.append(row)
    return np.array(cost_matrix)

# Read input from user
cost_matrix = get_user_input()

# Get optimal assignment
assignment, total_cost = hungarian_algorithm(cost_matrix)

# Output the results
print("Optimal Assignment:")
for agent, task in assignment:
    print(f"Agent {agent} -> Task {task}")
print("Total Cost:", total_cost)
```
