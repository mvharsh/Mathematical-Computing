# **SOLVING LINEAR PROGRAMMING PROBLEM (LPP) USING SIMPLEX METHOD**

**AIM:**

Solving a Linear Programming Problem (LPP) Using the Simplex Method and determines the optimal solution for maximizing or minimizing the given objective function.


**Step 1: Import Necessary Libraries**

  -	Import numpy for numerical operations.

  -	Import matplotlib.pyplot for visualization (though not explicitly used here).
  
  -	Import linprog from scipy.optimize for solving the linear programming problem.

**Step 2: Define the Function to Solve LPP**

  -	Accept the objective function coefficients (c), constraint matrix (A), and right-hand side (b).

  -	Convert a maximization problem into a minimization problem by negating c.

  -	Use the linprog function with the "highs" method to solve the problem.

  -	Return the optimal solution and optimal value if the problem is successfully solved. Otherwise, return an error message.

**Step 3: Get User Input**

  -	Accept the number of variables and constraints.

  -	Take input for the coefficients of the objective function.

  -	Take input for the constraint coefficients and the right-hand side values.

  -	Ask whether the user wants to maximize the objective function.

**Step 4: Solve the LPP and Display Results**

  -	Pass the collected inputs to the solve_lpp function.

  -	Display the optimal solution and value (or an error message if optimization fails).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import linprog

def solve_lpp(c, A, b, maximize=True):
    if maximize:
        c = [-ci for ci in c]  # Convert max to min by negating c

    result = linprog(c, A_ub=A, b_ub=b, method='highs')

    if result.success:
        optimal_value = -result.fun if maximize else result.fun
        return {"Optimal Solution": result.x, "Optimal Value": optimal_value}
    else:
        return {"Status": "Optimization failed", "Message": result.message}

# Get input from user
if __name__ == "__main__":
    num_vars = int(input("Enter the number of variables: "))
    num_constraints = int(input("Enter the number of constraints: "))

    print("Enter the coefficients of the objective function:")
    c = [float(input(f"c[{i+1}]: ")) for i in range(num_vars)]

    A = []
    b = []
    for i in range(num_constraints):
        print(f"Enter coefficients for constraint {i+1}:")
        row = [float(input(f"A[{i+1}][{j+1}]: ")) for j in range(num_vars)]
        A.append(row)
        b.append(float(input(f"Enter RHS value for constraint {i+1}: ")))

    maximize = input("Maximize the objective function? (yes/no): ").strip().lower() == "yes"

    solution = solve_lpp(c, A, b, maximize=maximize)
    print(solution)
```
