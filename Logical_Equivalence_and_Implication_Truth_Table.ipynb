# **IMPLEMENTATION OF TRUTH TABLES FOR LOGICAL EQUIVALENCE AND IMPLICATIONS**





**AIM:**

To construct and analyze truth tables for logical equivalence and implications using Python programming.

**ALGORITHM**

1.	Define individual functions for each logical operation (AND, OR, NOT, XOR, NAND, NOR, XNOR).
2.	Define functions for logical equivalence (A ⇔ B) and logical implication (A ⇒ B).
3.	Create a main function to construct truth tables for each logical operation.
4.	Use nested loops to iterate over all combinations of truth values (0 and 1) for two variables, A and B.
5.	For each combination of inputs:
a.	Compute the result of the logical operation using the corresponding function.
b.	Print the inputs and their corresponding output in a formatted table.
6.	Repeat the process for each logical operation in separate tables.
7.	Execute the main function to display all truth tables.
8.	Run the program and verify the outputs.


```python
def logical_and(A, B):
    return A and B

def logical_or(A, B):
    return A or B

def logical_not(A):
    return not A

def logical_xor(A, B):
    return A ^ B

def logical_nand(A, B):
    return not (A and B)

def logical_nor(A, B):
    return not (A or B)

def logical_xnor(A, B):
    return not (A ^ B)

# Function to evaluate logical equivalence (A ⇔ B)
def logical_equivalence(A, B):
    return A == B

# Function to evaluate logical implication (A ⇒ B)
def logical_implication(A, B):
    return not A or B

def construct_truth_tables():
    # Truth values: 1 (True) and 0 (False)
    truth_values = [1, 0]

    # AND Gate
    print("Truth Table for AND Gate")
    print(f"{'A':<5} {'B':<5} {'A AND B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {logical_and(A, B):<10}")
    print()

    # OR Gate
    print("Truth Table for OR Gate")
    print(f"{'A':<5} {'B':<5} {'A OR B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {logical_or(A, B):<10}")
    print()

    # NOT Gate
    print("Truth Table for NOT Gate")
    print(f"{'A':<5} {'NOT A':<10}")
    print("-" * 15)
    for A in truth_values:
        print(f"{A:<5} {int(logical_not(A)):<10}")
    print()

    # XOR Gate
    print("Truth Table for XOR Gate")
    print(f"{'A':<5} {'B':<5} {'A XOR B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {logical_xor(A, B):<10}")
    print()

    # NAND Gate
    print("Truth Table for NAND Gate")
    print(f"{'A':<5} {'B':<5} {'A NAND B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {int(logical_nand(A, B)):<10}")
    print()

    # NOR Gate
    print("Truth Table for NOR Gate")
    print(f"{'A':<5} {'B':<5} {'A NOR B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {int(logical_nor(A, B)):<10}")
    print()

    # XNOR Gate
    print("Truth Table for XNOR Gate")
    print(f"{'A':<5} {'B':<5} {'A XNOR B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {int(logical_xnor(A, B)):<10}")
    print()

    # Logical Implication
    print("Truth Table for Logical Implication")
    print(f"{'A':<5} {'B':<5} {'A ⇒ B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {int(logical_implication(A, B)):<10}")
    print()

    # Logical Equivalence
    print("Truth Table for Logical Equivalence")
    print(f"{'A':<5} {'B':<5} {'A ⇔ B':<10}")
    print("-" * 20)
    for A in truth_values:
        for B in truth_values:
            print(f"{A:<5} {B:<5} {int(logical_equivalence(A, B)):<10}")
    print()

# Main function
if __name__ == "__main__":
    print("Truth Tables for Logic Gates and Logical Statements")
    construct_truth_tables()
```

```python
def implies(p, q):
    return (not p) or q

# Loop through all combinations of p, q, r
print(f"{'P':<5}{'Q':<5}{'R':<5}{'Result'}")
print("-" * 25)
for p in [0, 1]:
    for q in [0, 1]:
        for r in [0, 1]:
            left = implies(p, q) and implies(q, r)
            right = implies(p, r)
            expression = implies(left, right)
            print(f"{p:<5}{q:<5}{r:<5}{int(expression)}")
```
